```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(10)
X = np.random.uniform(-2, 2, (100, 1))
Y = 2 * (X**2) + 3 * X + 5 + np.random.normal(0, 1, (100, 1))

X_poly = np.hstack((np.ones((100, 1)), X, X**2))
X_scaled = X_poly.copy()
X_scaled[:, 1] = (X_poly[:, 1] - np.mean(X_poly[:, 1])) / np.std(X_poly[:, 1])
X_scaled[:, 2] = (X_poly[:, 2] - np.mean(X_poly[:, 2])) / np.std(X_poly[:, 2])

m = len(Y)
epochs = 150
lr = 0.01
eps = 1e-8
beta1 = 0.9
beta2 = 0.999

theta = np.zeros((3, 1))
m_t = np.zeros((3, 1))
v_t = np.zeros((3, 1))
loss_history = []

for epoch in range(epochs):
    predictions = np.dot(X_scaled, theta)
    error = predictions - Y
    gradient = (1/m) * np.dot(X_scaled.T, error)
    
    m_t = beta1 * m_t + (1 - beta1) * gradient
    v_t = beta2 * v_t + (1 - beta2) * (gradient ** 2)
    
    m_t_corrected = m_t / (1 - beta1 ** (epoch + 1))
    v_t_corrected = v_t / (1 - beta2 ** (epoch + 1))
    
    theta = theta - (lr / (np.sqrt(v_t_corrected) + eps)) * m_t_corrected
    
    mse = (1 / (2 * m)) * np.sum(error ** 2)
    loss_history.append(mse)

plt.plot(loss_history, color='purple', label='Adam Loss')
plt.title('Adam Cost Over Epochs')
plt.xlabel('Epochs')
plt.ylabel('MSE')
plt.legend()
plt.grid(True)
plt.show()
```
