```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(10)
X = np.random.uniform(-2, 2, (100, 1))
Y = 2 * (X**2) + 3 * X + 5 + np.random.normal(0, 1, (100, 1))

X_poly = np.hstack((np.ones((100, 1)), X, X**2))
X_scaled = X_poly.copy()
X_scaled[:, 1] = (X_poly[:, 1] - np.mean(X_poly[:, 1])) / np.std(X_poly[:, 1])
X_scaled[:, 2] = (X_poly[:, 2] - np.mean(X_poly[:, 2])) / np.std(X_poly[:, 2])

m = len(Y)
epochs = 150
lr = 0.01
beta = 0.9
batch_size = 10

theta = np.zeros((3, 1))
v = np.zeros((3, 1))
loss_history = []

for epoch in range(epochs):
    for i in range(0, m, batch_size):
        xi = X_scaled[i:i+batch_size, :]
        yi = Y[i:i+batch_size, :]
        
        pred = np.dot(xi, theta)
        err = pred - yi
        gradient = (1/len(yi)) * np.dot(xi.T, err)
        
        v = beta * v + lr * gradient
        theta = theta - v
        
    full_pred = np.dot(X_scaled, theta)
    mse = (1 / (2 * m)) * np.sum((full_pred - Y) ** 2)
    loss_history.append(mse)

plt.plot(loss_history, color='green', label='Mini-Batch GD + Momentum Loss')
plt.title('Mini-Batch Gradient Descent Cost Over Epochs')
plt.xlabel('Epochs')
plt.ylabel('MSE')
plt.legend()
plt.grid(True)
plt.show()
```
